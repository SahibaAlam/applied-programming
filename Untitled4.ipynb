```python
import numpy as np

# Define the matrices
X = np.array([
    [3, 6],
    [5, 10],
    [10, 23]
])

Y = np.array([22, 45, 50])

# Display the matrices
print("X Matrix:")
print(X)
print("\nY Matrix:")
print(Y)
```

```python
lambda_value = 0

I = np.eye(2)

beta = np.linalg.inv(
    X.T @ X + lambda_value * I
) @ X.T @ Y

print("Linear Regression Coefficients:")
print(beta)
```

```python
lambda_value = 3

I = np.eye(2)

beta_ridge = np.linalg.inv(
    X.T @ X + lambda_value * I
) @ X.T @ Y

print("Ridge Regression Coefficients:")
print(beta_ridge)
```

```python
print("Linear Regression (Lambda = 0):", beta)
print("Ridge Regression (Lambda = 3):", beta_ridge)
```
